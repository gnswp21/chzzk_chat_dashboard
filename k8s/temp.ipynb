```python
import socket

api_server = "A855BA3AB61D52F9AE88C626F500DD13.gr7.ap-northeast-2.eks.amazonaws.com"  # 실제 EKS API 서버 도메인
port = 6443

s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
s.settimeout(5)  # 5초 타임아웃 설정
try:
    s.connect((api_server, port))
    print("연결 성공: API 서버에 접속되었습니다!")
except Exception as e:
    print("연결 실패:", e)
finally:
    s.close()
```
